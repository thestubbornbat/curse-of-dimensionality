```python
import math
```

```python
def volume_of_n_sphere(n, r):
    return (math.pi ** (n / 2) * r ** n) / math.gamma(n / 2 + 1)


print(volume_of_n_sphere(2, 1))
```

```python
from cgitb import small
import numpy as np

dimensions = np.arange(1,300,1)

ball_volumes = [volume_of_n_sphere(d, 1) for d in dimensions]

shell_thickness = 0.01

shell_volumes = []
for i in range(len(dimensions)):
    smaller_ball_vol = volume_of_n_sphere(dimensions[i], 1 - shell_thickness)
    shell_vol = ball_volumes[i] - smaller_ball_vol

    shell_volumes.append(shell_vol)


print(ball_volumes)
print(shell_volumes)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(dimensions, ball_volumes)
```

```python
plt.plot(dimensions, shell_volumes)
```

```python
volume_ratios = [x/y for x,y in zip(shell_volumes, ball_volumes)]
```

```python
plt.plot(dimensions, volume_ratios)
```
